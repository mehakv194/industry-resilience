# Agglomerative clustering — grouping robustness

Compare hierarchical groupings with K-Means using adjusted Rand index; agreement is stability, not ground truth.

**Real data only.** Run ingestion first. Never interpret an empty or skipped model as a finding.

```python
from pathlib import Path
import sys
for loc in [Path.cwd(),Path.cwd()/'notebooks',Path.cwd()/'research_expansion'/'notebooks']:
    if (loc/'common.py').exists():
        sys.path.insert(0,str(loc.resolve()));break
from common import root, load_panel, load_cases, output
```

```python
from sklearn.cluster import AgglomerativeClustering,KMeans
from sklearn.metrics import adjusted_rand_score,silhouette_score
from sklearn.preprocessing import StandardScaler
import pandas as pd
cols=['delta_pp','shock_std_yoy','negative_share','worst_yoy','post_mean_yoy']
df=load_cases().dropna(subset=cols).copy()
if len(df)<10:raise ValueError('Need 10+ observed cases')
X=StandardScaler().fit_transform(df[cols]);k=min(3,len(df)-1)
a=AgglomerativeClustering(n_clusters=k,linkage='ward').fit_predict(X)
b=KMeans(n_clusters=k,n_init=10,random_state=42).fit_predict(X)
df['hier_cluster']=a
output('hierarchical',{'model':'AgglomerativeWard','status':'fitted','n_cases':len(df),'k':k,
'adjusted_rand_vs_kmeans':adjusted_rand_score(a,b),
'silhouette':silhouette_score(X,a) if len(set(a))>1 else None,
'assignments':df[['country','industry_code','event','hier_cluster']].to_dict('records'),
'limitation':'Comparing two unsupervised methods does not establish objectively true categories.'})
display(df[['country','industry_code','event','hier_cluster']].head(12))
```
