# Isolation Forest — unusual historical cases

Identify observed shock-profile outliers; flags are NOT risk probabilities.

**Real data only.** Run ingestion first. Never interpret an empty or skipped model as a finding.

```python
from pathlib import Path
import sys
for loc in [Path.cwd(),Path.cwd()/'notebooks',Path.cwd()/'research_expansion'/'notebooks']:
    if (loc/'common.py').exists():
        sys.path.insert(0,str(loc.resolve()));break
from common import root, load_panel, load_cases, output
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
import pandas as pd
cols=['delta_pp','shock_std_yoy','negative_share','worst_yoy','post_mean_yoy']
df=load_cases().dropna(subset=cols).copy()
if len(df)<20:raise ValueError('Need at least 20 observed cases for an outlier detector')
X=StandardScaler().fit_transform(df[cols])
m=IsolationForest(n_estimators=150,contamination=0.1,random_state=42)
df['anomaly_flag']=(m.fit_predict(X)==-1);df['anomaly_score']=m.decision_function(X)
outliers=df.loc[df.anomaly_flag,['country','industry_code','event','anomaly_score']]
display(outliers)
output('isolation_forest',{'model':'IsolationForest','status':'fitted','n_cases':len(df),
'outlier_count':len(outliers),'outliers':outliers.to_dict('records'),
'limitation':'Isolation Forest marks unusual observed profiles; contamination=0.1 is a modeling choice, not a measured industry failure rate.'})
```
